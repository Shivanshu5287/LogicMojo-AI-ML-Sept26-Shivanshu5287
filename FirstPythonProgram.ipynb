```python
#libraries
from datetime import datetime
from rich import print

# Loop for the first name
while True:
    first_name = input("Please enter your first name: ").strip().title()
    if first_name.isalpha():
        break
    print("Please enter only alphabets in the first name field.")

# Loop for the middle name
while True:
    middle_name = input("Please enter your middle name: ").strip().title()
    if middle_name.isalpha():
        if middle_name.upper() == "NA":
            middle_name = middle_name.upper().replace("NA","")
        break
    print("Please enter only alphabets in the middle name field or NA if not applicable.")
    
# Loop for the last name
while True:
    last_name = input("Please enter your last name: ").strip().title()
    if last_name.isalpha():
        break
    print("Please enter only alphabets in the last name field.")
    
# Loop for the Date of birth
while True:
    dob = input("Please enter your date of birth in DD/MM/YYYY format: ").strip()
    try:
        valid_date = datetime.strptime(dob,"%d/%m/%Y").date()
        if valid_date > datetime.today().date():
            print(f"Date of birth cannot be in future.Please try again!")
            continue
        break
    except ValueError:
        print (f"Please enter the date in DD/MM/YYYY format only!")
        
# Loop for the PAN CARD Number
while True:
    pancard_number = input("Please enter your Pancard number: ").strip().upper()
    if pancard_number.isalnum():
        pancard_number = pancard_number.replace(pancard_number[3:],"XXXXXXX")
        break
    print("Please enter Valid Pancard Number only.")

# Loop for the Aadhaar Card Number
while True:
    aadharcard_number = input("Please enter your Aadhar card number: ").strip()
    if aadharcard_number.isdigit() and len(aadharcard_number) == 12:
        aadharcard_number = aadharcard_number.replace(aadharcard_number[:8],"XXXXXXXX")
        break
    print("Please enter Valid Aadhar Card Number only.")

# Loop for the Email address
while True:
    email_id = input("Please enter your Email Address: ").strip()
    if email_id.upper() == "NA" or ("@" in email_id and email_id.endswith(".com")):
        break
    print("Please enter Valid Email address only or enter NA.")

age = datetime.today().date().year - valid_date.year

print(f"my dob is {valid_date}")

print(f"""I [bold]{first_name}[/bold] [bold]{middle_name}[/bold] [bold]{last_name}[/bold],
born on [bold]{valid_date.strftime("%d/%m/%Y")}[/bold], [bold]{age}[/bold] years old. 
Holder of PAN card number [bold]{pancard_number}[/bold] and Aadhaar card number [bold]{aadharcard_number}[/bold].
I am a proud student of LogicMojo. 
You can reach out to me at [bold]{email_id}[/bold] if you need any feedback.""")
```
